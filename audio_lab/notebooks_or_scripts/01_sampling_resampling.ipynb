# 01 — Sampling and resampling

Load a recording, check its sample rate, and resample 16 kHz down to 8 kHz (phone quality).
Put your recording at `audio_lab/recordings/sample.wav` first, or change `INPUT_FILE` below.

```python
print("sahil")
```

```python
import os

import librosa
import numpy as np
import soundfile as sf

# ---- constants ----
INPUT_FILE = "../recordings/sample.wav"
OUTPUT_DIR = "../outputs"
TARGET_SR_HIGH = 16000
TARGET_SR_PHONE = 8000

os.makedirs(OUTPUT_DIR, exist_ok=True)

# ---- load original ----
audio_native, sr_native = librosa.load(INPUT_FILE, sr=None)  # sr=None keeps the file's own rate
print(f"File: {INPUT_FILE}")
print(f"Native sample rate: {sr_native} Hz")
print(f"Duration: {len(audio_native) / sr_native:.2f} s")
print(f"Samples: {len(audio_native)}")
print(f"dtype: {audio_native.dtype}, min={audio_native.min():.3f}, max={audio_native.max():.3f}")

# ---- resample to 16 kHz (typical STT input) ----
audio_16k = librosa.resample(audio_native, orig_sr=sr_native, target_sr=TARGET_SR_HIGH)
path_16k = os.path.join(OUTPUT_DIR, "sample_16k.wav")
sf.write(path_16k, audio_16k, TARGET_SR_HIGH)
print(f"\nSaved {path_16k}: {len(audio_16k)} samples at {TARGET_SR_HIGH} Hz")

# ---- resample to 8 kHz (phone quality / narrowband) ----
audio_8k = librosa.resample(audio_native, orig_sr=sr_native, target_sr=TARGET_SR_PHONE)
path_8k = os.path.join(OUTPUT_DIR, "sample_8k.wav")
sf.write(path_8k, audio_8k, TARGET_SR_PHONE)
print(f"Saved {path_8k}: {len(audio_8k)} samples at {TARGET_SR_PHONE} Hz")

# ---- what changed ----
print("\n--- Nyquist check ---")
print(f"16 kHz clip can represent frequencies up to {TARGET_SR_HIGH / 2:.0f} Hz")
print(f"8 kHz clip can represent frequencies up to {TARGET_SR_PHONE / 2:.0f} Hz")
print("Anything above the 8 kHz clip's limit (mostly consonant detail, sibilance) is gone.")
print("Listen to both files (Windows: open the .wav from the outputs folder) and compare.")
```
